# MoE presentation demo (adaptive-style comparison)

**One command** (recommended for slides):

```bash
python scripts/run_moe_presentation_demo.py
```

Outputs → `outputs/moe_presentation_demo/` (WAVs, PNG figures, `SUMMARY.md`, `demo_metrics.csv`).

**Adaptive reference** (run adaptive notebook demo first):
- `outputs/original_demo.wav`
- `outputs/oracle_adaptive_reconstruction.wav`
- `outputs/classifier_adaptive_reconstruction.wav`

This notebook re-runs the same pipeline and displays results inline (~2 min with checkpoints).

````python
import subprocess
import sys
from pathlib import Path

REPO = Path.cwd()
script = REPO / "scripts" / "run_moe_presentation_demo.py"
assert script.exists(), "Run from repo root"

subprocess.run([sys.executable, str(script)], check=True, cwd=REPO)
````

````python
from pathlib import Path
from IPython.display import Audio, Image, display
import pandas as pd

OUT = Path("outputs/moe_presentation_demo")
print((OUT / "SUMMARY.md").read_text())
display(pd.read_csv(OUT / "demo_metrics.csv"))
display(Image(filename=str(OUT / "figures" / "waveforms_comparison.png")))
display(Image(filename=str(OUT / "figures" / "routing_timeline.png")))
display(Image(filename=str(OUT / "figures" / "metrics_bar.png")))
````

````python
from IPython.display import Audio, display

wav = OUT / "wav"
sr = 16000
for name, label in [
    ("01_original.wav", "Clean"),
    ("02_mixed_noisy.wav", "Mixed noisy"),
    ("03_moe_oracle.wav", "MoE oracle route"),
    ("04_moe_classifier.wav", "MoE classifier (smoothed RF)"),
]:
    print(label)
    display(Audio(filename=str(wav / name), rate=sr))

ref = OUT / "adaptive_reference"
if (ref / "adaptive_classifier.wav").exists():
    print("\nAdaptive classifier (reference copy)")
    display(Audio(filename=str(ref / "adaptive_classifier.wav"), rate=sr))
````
