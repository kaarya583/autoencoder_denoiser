```python
# Resolve the repository when launched from notebooks/legacy or the repo root.
import os
import sys
from pathlib import Path

for candidate in (Path.cwd(), *Path.cwd().parents):
    if (candidate / "moe_baseline").is_dir() and (candidate / "README.md").is_file():
        os.chdir(candidate)
        if str(candidate) not in sys.path:
            sys.path.insert(0, str(candidate))
        break
```
